```python
def modify_arguments(immutable, mutable):
    immutable = immutable + 1
    mutable.append(4)
    print("Inside the function:")
    print("Immutable:", immutable)
    print("Mutable:", mutable)

immutable_arg = 10
mutable_arg = [1, 2, 3]

print("Before calling the function:")
print("Immutable:", immutable_arg)
print("Mutable:", mutable_arg)

modify_arguments(immutable_arg, mutable_arg)

print("After calling the function:")
print("Immutable:", immutable_arg)
print("Mutable:", mutable_arg)
```

```python
##### def basic_arithmetic(num1, num2):
    """Performs basic arithmetic operations on two numbers."""
    print(f"Addition: {num1 + num2}")
    print(f"Subtraction: {num1 - num2}")
    print(f"Multiplication: {num1 * num2}")
    if num2 != 0:  # Avoid division by zero
        print(f"Division: {num1 / num2}")
    else:
        print("Cannot divide by zero.")

# Get user input
try:
    num1 = float(input("Enter the first number: "))
    num2 = float(input("Enter the second number: "))
    basic_arithmetic(num1, num2)

except ValueError:
    print("Invalid input. Please enter numbers.")
```

```python
a = True
b = False

# Logical AND operator (and)
print("a and b:", a and b)  # Output: False

# Logical OR operator (or)
print("a or b:", a or b)   # Output: True

# Logical NOT operator (not)
print("not a:", not a)      # Output: False
print("not b:", not b)      # Output: True

# Combining logical operators
x = 5
y = 10
print((x > 0) and (y < 15)) # Output: True
print((x < 0) or (y > 20)) # Output: False
```

```python
def convert_string_to_types(input_string):
    """
    Converts a string to integer, float, and boolean types if possible.

    Args:
        input_string: The string to convert.

    Returns:
        A dictionary containing the converted values or None if conversion is not possible.
    """
    result = {}

    try:
        result['integer'] = int(input_string)
    except ValueError:
        result['integer'] = None

    try:
        result['float'] = float(input_string)
    except ValueError:
        result['float'] = None

    if input_string.lower() == 'true':
        result['boolean'] = True
    elif input_string.lower() == 'false':
        result['boolean'] = False
    else:
         result['boolean'] = None

    return result

# Get user input
user_input = input("Enter a value: ")

# Convert the input string
converted_values = convert_string_to_types(user_input)

# Print the results
if converted_values:
    print("Converted Values:")
    for data_type, value in converted_values.items():
        print(f"{data_type}: {value}")
else:
    print("No conversion possible.")
```

```python
def reverse_string(s):
    reversed_s = ""
    length = len(s) - 1
    while length >= 0:
        reversed_s += s[length]
        length -= 1
    return reversed_s

input_string = "Hello, World!"
reversed_string = reverse_string(input_string)
print(f"Original string: {input_string}")
print(f"Reversed string: {reversed_string}")
```

```python
number = int(input("Enter a non-negative integer: "))
factorial = 1

if number < 0:
    print("Factorial is not defined for negative numbers.")
elif number == 0:
    print("The factorial of 0 is 1")
else:
    while number > 0:
        factorial *= number
        number -= 1
    print("The factorial is", factorial)
```
